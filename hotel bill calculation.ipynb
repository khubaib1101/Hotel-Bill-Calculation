```python
# Hotel Room Booking 16

def calculate_hotel_bill(room_type,nights,breakfast=False):
  room_type= room_type.lower()
  if room_type=='standard':
    room_charge= 5000
  elif room_type == 'deluxe':
    room_charge = 8000
  elif room_type == 'suite':
    room_charge= 1500
  else:
    print("Enter proper room type")

  if nights>=7:
    discount = room_charge * 0.15
  elif nights>3 and  nights <=6:
    discount = room_charge * 0.10
  else:
    discount = 0
  total_bill = room_charge - discount
  if breakfast==True:
    total_bill+=1000*nights

  print("your bill is-->=",total_bill)
calculate_hotel_bill('suite',7,breakfast=False)

```
